### Task 10: Wasserstein GAN with Gradient Penalty (WGAN-GP)

```python
#task10
import torch
import torch.nn as nn

# Real Images
real = torch.randn(4,100)

# Fake Images
fake = torch.randn(4,100)

# Generator
class Generator(nn.Module):

    def __init__(self):
        super().__init__()

        self.fc = nn.Linear(100,100)

    def forward(self,x):
        return torch.tanh(self.fc(x))

# Critic
class Critic(nn.Module):

    def __init__(self):
        super().__init__()

        self.fc = nn.Linear(100,1)

    def forward(self,x):
        return self.fc(x)

generator = Generator()
critic = Critic()

# Generate Fake Images
noise = torch.randn(4,100)
fake_images = generator(noise)

# Critic Scores
real_score = critic(real)
fake_score = critic(fake_images)

# Wasserstein Loss
wasserstein_loss = torch.mean(fake_score) - torch.mean(real_score)

# Gradient Penalty (Simple)
gradient_penalty = ((real - fake_images)**2).mean()

# Total Loss
critic_loss = wasserstein_loss + 10 * gradient_penalty

print("Wasserstein Loss :", wasserstein_loss.item())
print("Gradient Penalty :", gradient_penalty.item())
print("Critic Loss :", critic_loss.item())
```
