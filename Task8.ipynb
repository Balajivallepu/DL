### Task 8: Semantic Segmentation with Custom Dice Loss

```python
#task8
import torch
import torch.nn as nn

# Sample Image and Mask
X = torch.randn(2,1,64,64)
Y = torch.randint(0,2,(2,1,64,64)).float()

# Simple U-Net
class UNet(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Conv2d(1,16,3,padding=1)
        self.decoder = nn.Conv2d(16,1,3,padding=1)

    def forward(self,x):

        x = torch.relu(self.encoder(x))
        x = torch.sigmoid(self.decoder(x))

        return x

model = UNet()

# Prediction
pred = model(X)

# Dice Loss
def dice_loss(pred,target):

    smooth = 1

    intersection = (pred*target).sum()

    dice = (2*intersection + smooth)/(pred.sum()+target.sum()+smooth)

    return 1-dice

# BCE Loss
bce = nn.BCELoss()

loss1 = bce(pred,Y)
loss2 = dice_loss(pred,Y)

total_loss = loss1 + loss2

print("BCE Loss :",loss1.item())
print("Dice Loss :",loss2.item())
print("Total Loss :",total_loss.item())
```
