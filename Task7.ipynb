### Task 7: ResNet-50 Style Residual Bottleneck Block


```python
#task7
import torch
import torch.nn as nn

# Residual Bottleneck Block
class Bottleneck(nn.Module):

    def __init__(self):
        super().__init__()

        # 1x1 Convolution
        self.conv1 = nn.Conv2d(3,16,1)

        # Depthwise Convolution
        self.depthwise = nn.Conv2d(16,16,3,padding=1,groups=16)

        # Pointwise Convolution
        self.pointwise = nn.Conv2d(16,16,1)

        # Output Convolution
        self.conv2 = nn.Conv2d(16,3,1)

        self.relu = nn.ReLU()

    def forward(self,x):

        skip = x

        x = self.relu(self.conv1(x))
        x = self.relu(self.depthwise(x))
        x = self.relu(self.pointwise(x))
        x = self.conv2(x)

        x = x + skip      # Residual Connection

        return self.relu(x)

# Create Model
model = Bottleneck()

# Dummy Input
X = torch.randn(1,3,64,64)

# Forward Pass
output = model(X)

print("Input Shape :", X.shape)
print("Output Shape:", output.shape)

# Parameter Count
params = sum(p.numel() for p in model.parameters())
print("Total Parameters:", params)
```
